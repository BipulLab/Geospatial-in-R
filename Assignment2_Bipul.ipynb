**8.Create two set of Color**

```python
# Create two sets of colors
set1 = {"red", "blue", "green"}
set2 = {"yellow", "green", "blue", "black"}

# Find union (all unique elements from both sets)
union_set = set1.union(set2)

# Find intersection (common elements in both sets)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```

**9.Simple python function**

```python
def first_last_char(text):
    # Check if the string is not empty
    if len(text) > 0:
        print("First character:", text[0])
        print("Last character:", text[-1])
    else:
        print("The string is empty.")

# Example usage
first_last_char("Bangladesh")
```

**10. Python Function**

```python
def to_uppercase(text):
    result = text.upper()
    return result

# Example usage
print(to_uppercase("bangladesh"))
```
